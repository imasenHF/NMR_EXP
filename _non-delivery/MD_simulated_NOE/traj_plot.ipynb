```python
import numpy as np
import matplotlib.pyplot as plt

def lorentzian(x, x0, FWHM, y0):
    """
    计算洛伦兹展宽函数
    :param x: 横坐标，频率或其他相关变量
    :param x0: 洛伦兹分布的中心位置
    :param FWHM: 半高全宽
    :param y0: 最大高度
    :return: 洛伦兹展宽函数的值
    """
    return y0 * (FWHM / 2) / (np.pi * ((x - x0)**2 + (FWHM / 2)**2))

def gaussian(x, x0, FWHM, y0):
    """
    计算高斯展宽函数
    :param x: 横坐标，频率或其他相关变量
    :param x0: 高斯分布的中心位置
    :param FWHM: 半高全宽
    :param y0: 最大高度
    :return: 高斯展宽函数的值
    """
    return y0 * np.exp(-4 * np.log(2) * (x - x0)**2 / FWHM**2)

if __name__ == "__main__":
    # 参数设置
    x_positions = np.array([0, 1, 2, 3, 4])  # 柱的位置
    y_heights = np.array([1, 0.8, 0.6, 0.9, 1.2])  # 每个柱的高度
    FWHM = 0.15  # 所有柱的统一半高全宽
    x = np.linspace(-1, 5, 1000)  # 横坐标范围，足够大以容纳所有柱
    
    # 初始化一个空的数组用于累加洛伦兹展宽结果
    y_total = np.zeros_like(x)
    
    # 对每个柱应用洛伦兹展宽，并累加结果
    for i in range(len(x_positions)):
        y_total += gaussian(x, x_positions[i], FWHM, y_heights[i])
    
    # 绘制图形
    plt.plot(x, y_total, label="Total Lorentzian Broadening")
    plt.title("Lorentzian Broadening of Multiple Bars with Unified FWHM")
    plt.xlabel('Position (x)')
    plt.ylabel('Intensity')
    plt.grid(True)
    plt.legend()
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 读取txt文件数据，假设文件名为data.txt
data = np.loadtxt('aza_N_traj.txt', dtype={'names': ('residue', 'x', 'y', 'z'),
                                     'formats': ('U10', 'f4', 'f4', 'f4')})

data2 = np.loadtxt('sys_O_traj.txt', dtype={'names': ('residue2', 'x2', 'y2', 'z2'),
                                     'formats': ('U10', 'f4', 'f4', 'f4')})

# 提取z坐标
position_N = data['z']# 柱的位置
position_O = data2['z2']
pop = 1  # 每个柱的高度
FWHM = 0.1  # 所有柱的统一半高全宽
x = np.linspace(-10, 85, 1000)  # 横坐标范围，足够大以容纳所有柱

# 初始化一个空的数组用于累加洛伦兹展宽结果
y_N = np.zeros_like(x)
for i in range(len(position_N)):
    y_N += lorentzian(x, position_N[i], FWHM, pop)

y_N /= (np.sum(y_N) * 95)
np.save('pop_N.npy', y_N)

y_O = np.zeros_like(x)
for i in range(len(position_O)):
    y_O += lorentzian(x, position_O[i], FWHM, pop)

y_O /= (np.sum(y_O) * 95)
np.save('pop_O.npy', y_O)

# 绘制散点图
plt.figure(figsize=(6.44/2.54, 6.44/2.54)) # 6.44
plt.rcParams['font.family'] = 'Arial'  # 设置全局字体为 Arial
ax = plt.gca()  # 设置图形边框线宽
for spine in ax.spines.values():
    spine.set_linewidth(1)  # 设置所有边框的线宽为 1

plt.plot(y_N, x, label="Nitrogen", linewidth=1)
plt.plot(y_O, x, label="Oxygen", linewidth=1)

# 添加标题和标签
plt.title('')
plt.xlabel('Population', fontsize=12, fontweight='semibold')
plt.ylabel('Height (nm)', fontsize=12, fontweight='semibold')
# plt.grid(True)
plt.ylim(-10, 85)
plt.legend(fontsize=8, loc='upper right', frameon=False)

plt.savefig('oxygen_nitrogen_population_lor.png', dpi=600, bbox_inches='tight')
plt.show()

```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# 读取氮原子数据
df = pd.read_csv("prod_nitrogen.txt", sep=r'\s+', header=None)
x = df[1]
y = df[2]
z = df[3]

# 选择投影平面：这里以 xz 为例
coord1 = x  # 横轴为 x
coord2 = z  # 纵轴为 z

# 设置二维网格用于密度估计
bins = 200
hist, xedges, yedges = np.histogram2d(coord1, coord2, bins=bins)

# 将每个点映射到网格索引
xidx = np.digitize(coord1, xedges) - 1
yidx = np.digitize(coord2, yedges) - 1
xidx = np.clip(xidx, 0, bins - 1)
yidx = np.clip(yidx, 0, bins - 1)

# 获取每个点所在格子的密度值
density = hist[xidx, yidx]

# 绘图
plt.figure(figsize=(6, 5))
plt.scatter(coord1, coord2, s=50, c=density, cmap='viridis', edgecolors='none')
plt.xlabel("x (Å)")
plt.ylabel("z (Å)")
plt.title("Nitrogen atom density projection on xz plane")
plt.grid(True, linestyle="--", linewidth=0.5, color="gray", alpha=0.6)
plt.tight_layout()
plt.savefig("nitrogen_density_xz.png", dpi=600, bbox_inches="tight")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 读取数据
df = pd.read_csv("prod_nitrogen.txt", sep=r'\s+', header=None)
x = df[1]
y = df[2]
z = df[3]

# 设置二维网格参数
bins = 150

# 计算 xz 密度
hist_xz, xedges_xz, zedges_xz = np.histogram2d(x, z, bins=bins)
xidx = np.clip(np.digitize(x, xedges_xz) - 1, 0, bins - 1)
zidx = np.clip(np.digitize(z, zedges_xz) - 1, 0, bins - 1)
density_xz = hist_xz[xidx, zidx]

# 计算 yz 密度
hist_yz, yedges_yz, zedges_yz = np.histogram2d(y, z, bins=bins)
yidx = np.clip(np.digitize(y, yedges_yz) - 1, 0, bins - 1)
zidx2 = np.clip(np.digitize(z, zedges_yz) - 1, 0, bins - 1)
density_yz = hist_yz[yidx, zidx2]

# 设置共享色阶
vmin = min(density_xz.min(), density_yz.min())
vmax = max(density_xz.max(), density_yz.max())

# 创建子图
fig, axs = plt.subplots(1, 2, figsize=(12, 5), sharey=True)
cmap = plt.cm.Spectral_r

# xz 投影图
sc1 = axs[0].scatter(x, z, s=30, c=density_xz, cmap=cmap, vmin=vmin, vmax=vmax, edgecolors='none')
axs[0].set_xlabel("x (Å)")
axs[0].set_ylabel("z (Å)")
axs[0].set_title("XZ Projection")
axs[0].grid(True, linestyle="--", linewidth=0.5, color="gray", alpha=0.6)

# yz 投影图
sc2 = axs[1].scatter(y, z, s=50, c=density_yz, cmap=cmap, vmin=vmin, vmax=vmax, edgecolors='none')
axs[1].set_xlabel("y (Å)")
axs[1].set_title("YZ Projection")
axs[1].grid(True, linestyle="--", linewidth=0.5, color="gray", alpha=0.6)

# 添加共享 colorbar
# cbar = fig.colorbar(sc2, ax=axs.ravel().tolist(), shrink=0.85, label="Density")

# plt.tight_layout()
plt.savefig("nitrogen_density_xz_yz.png", dpi=600, bbox_inches="tight")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ========== 子图1 & 子图2：xz / yz 投影密度散点图 ==========

df = pd.read_csv("prod_nitrogen.txt", sep=r'\s+', header=None)
x = df[1]
y = df[2]
z = df[3]
bins = 150

# xz 密度
hist_xz, xedges_xz, zedges_xz = np.histogram2d(x, z, bins=bins)
xidx = np.clip(np.digitize(x, xedges_xz) - 1, 0, bins - 1)
zidx = np.clip(np.digitize(z, zedges_xz) - 1, 0, bins - 1)
density_xz = hist_xz[xidx, zidx]

# yz 密度
hist_yz, yedges_yz, zedges_yz = np.histogram2d(y, z, bins=bins)
yidx = np.clip(np.digitize(y, yedges_yz) - 1, 0, bins - 1)
zidx2 = np.clip(np.digitize(z, zedges_yz) - 1, 0, bins - 1)
density_yz = hist_yz[yidx, zidx2]

# 统一色标范围
vmin = min(density_xz.min(), density_yz.min())
vmax = max(density_xz.max(), density_yz.max())
cmap = plt.cm.Spectral_r

# ========== 创建3列子图（统一大小，隐藏标题、colorbar）==========
fig, axs = plt.subplots(1, 3, figsize=(3.0 * 6.44 / 2.54, 6.44 / 2.54), sharey=False)
plt.rcParams['font.family'] = 'Arial'

# 子图1：xz
sc1 = axs[0].scatter(x, z, s=30, c=density_xz, cmap=cmap, vmin=vmin, vmax=vmax, edgecolors='none')
axs[0].set_xlabel("X (Å)", fontsize=12)
axs[0].set_ylabel("Z (Å)", fontsize=12)
axs[0].tick_params(labelsize=10)
axs[0].set_title("")
axs[0].grid(True, linestyle="--", linewidth=0.5, color="gray", alpha=0.6)

# 子图2：yz
sc2 = axs[1].scatter(y, z, s=30, c=density_yz, cmap=cmap, vmin=vmin, vmax=vmax, edgecolors='none')
axs[1].set_xlabel("Y (Å)", fontsize=12)
axs[1].set_ylabel("Z (Å)", fontsize=12)
axs[1].tick_params(labelsize=10)
axs[1].set_title("")
axs[1].grid(True, linestyle="--", linewidth=0.5, color="gray", alpha=0.6)

# ========== 子图3：Lorentz 拓展柱状密度图 ==========

def lorentzian(x, x0, gamma, A):
    return (A * (gamma / 2) ** 2) / ((x - x0) ** 2 + (gamma / 2) ** 2)

data = np.loadtxt('prod_nitrogen.txt', dtype={'names': ('residue', 'x', 'y', 'z'),
                                              'formats': ('U10', 'f4', 'f4', 'f4')})
data2 = np.loadtxt('prod_oxygen.txt', dtype={'names': ('residue2', 'x2', 'y2', 'z2'),
                                             'formats': ('U10', 'f4', 'f4', 'f4')})
position_N = data['z']
position_O = data2['z2']
pop = 1
FWHM = 0.15
x_lor = np.linspace(-10, 85, 1000)
y_N = np.zeros_like(x_lor)
for i in range(len(position_N)):
    y_N += lorentzian(x_lor, position_N[i], FWHM, pop)
y_N /= (np.sum(y_N) * 95)

y_O = np.zeros_like(x_lor)
for i in range(len(position_O)):
    y_O += lorentzian(x_lor, position_O[i], FWHM, pop)
y_O /= (np.sum(y_O) * 95)

# 子图3：线图
axs[2].plot(y_N, x_lor, label="Nitrogen", linewidth=1)
axs[2].plot(y_O, x_lor, label="Oxygen", linewidth=1)
axs[2].set_xlabel('Population', fontsize=12)
axs[2].set_ylabel('Z (Å)', fontsize=12)
axs[2].tick_params(labelsize=10)
axs[2].set_ylim(-10, 85)
axs[2].set_title("")
axs[2].legend(fontsize=8, loc='upper right', frameon=False)

# 去掉 colorbar
# fig.colorbar(...) 不添加

# 保存
plt.tight_layout()
plt.savefig("three_panel_projection_density_2.png", dpi=600, bbox_inches="tight")
plt.show()
```
